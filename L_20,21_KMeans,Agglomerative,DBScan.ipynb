```python
from sklearn.datasets import load_iris
data = load_iris()
x = data.data
```

```python
from sklearn.cluster import KMeans
km = KMeans(3)
```

```python
km.fit(x)
```

```python
centroid = km.cluster_centers_
label = km.labels_
```

```python
label
```

```python
import pandas as pd
df = pd.DataFrame(x, columns=data.feature_names)

df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:, 0], x[:, 1], c=label)
plt.scatter(centroid[:, 0], centroid[:, 1], marker='*', c='red')
plt.show()
```

```python
df['cluster'] = label
df.head()
```

```python
import matplotlib.pyplot as plt
plt.scatter(x[:, 0], x[:, 1], c=data.target)
plt.scatter(centroid[:, 0], centroid[:, 1], marker='*', c='red')
plt.show()
```

```python
data.target
```

```python
import sklearn
sklearn.metrics.confusion_matrix(data.target, df['cluster'])
```

```python
sklearn.metrics.silhouette_score(x, label)
# it shows how well it has clustered
```

```python
from sklearn.cluster import DBSCAN
import matplotlib.pyplot as plt
model = DBSCAN(eps=0.5, min_samples=5)

model.fit(x)
```

```python
n_cluster = len(set(model.labels_))
n_cluster
```

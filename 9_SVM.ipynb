```python
import numpy as np
from sklearn.datasets import make_classification
from sklearn.svm import SVC
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import hinge_loss

# Step 1: Generate synthetic binary classification data
X, y = make_classification(n_samples=200, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1, random_state=42)

# Convert labels to {-1, 1} for hinge loss
y_transformed = np.where(y == 0, -1, 1)

# Step 2: Feature scaling
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)

# Step 3: Train SVM with linear kernel
svm = SVC(kernel='linear', C=1.0)
svm.fit(X_scaled, y_transformed)

# Step 4: Compute decision function and hinge loss
decision_values = svm.decision_function(X_scaled)

# Manual Hinge Loss
manual_hinge_loss = np.mean(np.maximum(0, 1 - y_transformed * decision_values))

# Or use scikit-learn's built-in
sklearn_hinge_loss = hinge_loss(y_transformed, decision_values)

print(f"Manual Hinge Loss: {manual_hinge_loss:.4f}")
print(f"sklearn Hinge Loss: {sklearn_hinge_loss:.4f}")
```
